**Question:** How would you structure the data to observe trends in the average valuation of companies from 2020 to 2021?  

```python

# create one sub_set
years_to_filter = [2020, 2021]
companies_filtered = companies[companies['Date Joined'].dt.year.isin(years_to_filter)]

# Create function transform
def clean_valuation(value):
    value = str(value).replace('$', '').strip()
    
    if 'B' in value:
        # Multiplicamos por MIL MILLONES
        return int(float(value.replace('B', '')) * 1_000_000_000)
    elif 'M' in value:
        # Multiplicamos por UN MILLÓN
        return int(float(value.replace('M', '')) * 1_000_000)
    
    return int(float(value))

# Trasform valuation to int format
companies_filtered['Valuation'] = companies_filtered['Valuation'].apply(clean_valuation)
# Extract year of Year Joined
companies_filtered['Year Joined'] = companies_filtered['Date Joined'].dt.to_period('Y')
# Group for year and calculate mean for year
trend_data = companies_filtered.groupby('Year Joined')['Valuation'].mean()

plt.bar(trend_data.index.astype(str), trend_data.values, color=['#3498db','#e67e22'])
plt.title('Promedio de Valuación de empresas unicornio por año', fontsize=14, fontweight="bold")
plt.xlabel('Año de empresa unicornio',fontsize=12)    
plt.ylabel('Valuación promedio',fontsize=12)
    
plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f'${x/1e9:.1f}B'))
for i,v in enumerate(trend_data.values):
    plt.text(i,v+(v*0.02), f"${v/1e9:.1f}B", ha='center', fontweight='bold')
    
plt.tight_layout()
plt.savefig('barras_valuation.png')
```

### Compare trends over time

Implement the structuring approach that you have identified in order to compare trends over time in the average valuation of companies that became unicorns in the year you selected above and in another year of your choice. Keep in mind the data type of the `Valuation` column and what the values in that column contain currently.

```python
# After identifying the additional year and time interval of interest, proceed with the following:
# Search companies for year 
#companies['Date Joined'].value_counts()
# select the range for this example 2015 to 2021
year=2021
# Step 1. Filter by the additional year to create a subset that consists of companies that joined in that year.
subset_companies=companies[companies['Date Joined'].dt.year==year].head()                        
# Step 2. Concatenate that new subset with the subset that you defined previously.
subset_companies=pd.concat([companies_subset,subset_companies])
subset_companies = subset_companies.reset_index(drop=True)
subset_companies.head()
# Step 3. As needed, add a column that contains the time interval that each data point belongs to, 
#         in the concatenated DataFrame.
subset_companies['Year_Joined']=subset_companies['Date Joined'].dt.year
# Step 4. Transform the `Valuation` column as needed.
subset_companies['Valuation'] = subset_companies['Valuation'].apply(clean_valuation)


# Step 5. Group by the time interval.
#         Aggregate by computing average valuation of companies that joined per interval of the corresponding year.
#         Save the resulting DataFrame in a new variable.

# For acquire Quarter I use dt.to_period('Q')
# dt.to period : maintains the period
# dt.strftime : format the quartiles %Y format years of 4 numbers (2015) Q%q format count quartile every 4 months
subset_companies['Quarter_Joined'] = subset_companies['Date Joined'].dt.to_period('Q').dt.strftime('%Y-Q%q')

# trends_quarterly : Give a columns Quarter_Joined and calculate the mean of Valuation for quarterly
# reset_index help to transform 'Quarter_Joined' in index to trends_quarterly
# rename the column 'Valuation' to 'Average_Valuation' because now we have the mean of Valuation for quarterly
trends_quarterly = (
    subset_companies
    .groupby('Quarter_Joined')['Valuation']
    .mean()
    .reset_index()
    .rename(columns={'Valuation': 'Average_Valuation'})
)
# Display the first few rows of the new DataFrame to confirm that it was created.
trends_quarterly.head()
### YOUR CODE HERE ###


```

### Visualize the average valuation over the quarters

Using the subset of companies that joined in the years of interest, create a grouped bar plot to visualize the average valuation over the quarters, with two bars for each time interval. There will be two bars for each time interval. This allows you to compare quarterly values between the two years.

```python
# Extract the years and quartile
subset_recent['Year_Only'] = subset_recent['Date Joined'].dt.year
subset_recent['Quarter_Number'] = subset_recent['Date Joined'].dt.quarter

# I format to valuation
subset_recent['Valuation'] = subset_recent['Valuation'].apply(clean_valuation)

# I grouped for year and quartile and valuation mean
grouped = subset_recent.groupby(['Year_Only', 'Quarter_Number'])['Valuation'].mean().reset_index()
# Now finally I generate bar graph, I compare the valuation average for quartile and year
plt.figure(figsize=(12, 6))
sns.barplot(
    data=grouped, # pass the grouped data
    x='Quarter_Number', # axis x be quartile 
    y='Valuation', # axis y is a average valuation
    hue='Year_Only', # information box for year mean valuation
    palette='Paired' # colors palette
)

plt.title('Comparación de Valuación Promedio por Trimestre')
plt.xlabel('Trimestre (Quarter)')
plt.ylabel('Valuación Promedio (Billones USD)')
plt.xticks(ticks=[0,1,2,3], labels=['Q1', 'Q2', 'Q3', 'Q4'])
plt.legend(title='Año de Unión')
plt.show()

```
