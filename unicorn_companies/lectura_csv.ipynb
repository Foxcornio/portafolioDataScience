## Determinar si los datos completos son similares.

Al verificar los datos que si estan completos en comparación con los que no lo estan podremos determinar si son similares o contienen errores.

```python
df_clean = df.dropna().drop_duplicates() # Eliminamos los datos con campos vacios y duplicados
# Transformamos los datos de fondos
parse_column = df_clean['Funding'].apply(data_format.parse_financial)
df_clean['Funding'] = pd.to_numeric(parse_column, errors='coerce').astype('Int64')

# Transformamos los datos de fondos
parse_column = df_clean['Valuation'].apply(data_format.parse_financial)
df_clean['Valuation'] = pd.to_numeric(parse_column, errors='coerce').astype('Int64')
df_clean['Date Joined'] = pd.to_datetime(df['Date Joined']) # Cambiamos a date_time
df_clean['Year Founded'] = pd.to_datetime(df['Year Founded']) # cambiamos a date_time


# Pasamos directo a la grafica de barras.
companies_industry_valuation = df_clean.groupby('Industry')['Valuation'].sum()
companies_industry_valuation.sort_values(ascending=False)
valuation_dict = companies_industry_valuation.to_dict()

# En un diccionario, usamos keys() para las etiquetas y values() para los datos
label_x = list(valuation_dict.keys())
label_y = list(valuation_dict.values())
# Creamos el gráfico
plt.bar(label_x, label_y, color="purple")
plt.title('Valoración por industria')
plt.ylabel('Valoración en Billones',rotation=90)
plt.xticks(rotation=90)
plt.show()
```

Al Igual que las ocasiones anteriores vemos que la industria de finanzas domina el mercado.

Ahora hagamos lo mismo con los fondos.

```python
# Pasamos directo a la grafica de barras.
companies_industry_valuation = df_clean.groupby('Industry')['Funding'].sum()
companies_industry_valuation.sort_values(ascending=False)
valuation_dict = companies_industry_valuation.to_dict()

# En un diccionario, usamos keys() para las etiquetas y values() para los datos
label_x = list(valuation_dict.keys())
label_y = list(valuation_dict.values())
# Creamos el gráfico
plt.bar(label_x, label_y, color="purple")
plt.title('Valoración por industria')
plt.ylabel('Valoración en Billones',rotation=90)
plt.xticks(rotation=90)
plt.show()
```

Como puede observarse la industria con mayor numero de fondos es la de finanzas, pero ¿es así en todos los paises?

```python
df_clean["Industry"] = df_clean["Industry"].replace("Artificial Intelligence", "Artificial intelligence") # se identifica una variacion en Artificial intelligence y se estandariza
dict_country={} # se crea el diccionario que almacenara los datos
for (country, industry), group in df_clean.groupby(['Country/Region', 'Industry']):
    # Con un loop recorremos agrupando por pais y industria
    if country not in dict_country: # verificamos que no este el pais en el dict
        dict_country[country] = {}    # agregamos el pais al diccionario
    # Guardamos los datos de esa industria dentro del país
    dict_country[country][industry] = group.to_dict(orient="records")

import ipywidgets as widgets
from IPython.display import display, clear_output
import pandas as pd

# 1. Configurar los menús
menu_pais = widgets.Dropdown(
    options=sorted(dict_country.keys()),
    value='Mexico',
    description='País:',
)

# Las opciones de industria deben depender del país inicial
menu_industria = widgets.Dropdown(
    options=sorted(dict_country[menu_pais.value].keys()),
    value='Fintech', 
    description='Industria:',
)

salida = widgets.Output()

# 2. Función que lee AMBOS menús al mismo tiempo
def actualizar_tabla(change):
    pais = menu_pais.value
    industria = menu_industria.value
    
    with salida:
        clear_output(wait=True)
        try:
            # Accedemos directamente al diccionario con ambos valores
            datos = dict_country[pais][industria]
            df_final = pd.DataFrame(datos)
            print(f"Resultados para {pais} en la industria {industria}:")
            display(df_final)
        except KeyError:
            print(f"No hay datos para {industria} en {pais}.")

# 3. Función especial para actualizar las industrias cuando cambias de país
def actualizar_menu_industrias(change):
    nuevo_pais = change['new']
    # Actualizamos las opciones del segundo menú según el país
    menu_industria.options = sorted(dict_country[nuevo_pais].keys())
    actualizar_tabla(None)

# 4. Conectar los eventos
menu_pais.observe(actualizar_menu_industrias, names='value')
menu_industria.observe(actualizar_tabla, names='value')

# 5. Mostrar
display(menu_pais, menu_industria, salida)
actualizar_tabla(None)
```

Se identifico que Artificial intelligence tenia una variación
