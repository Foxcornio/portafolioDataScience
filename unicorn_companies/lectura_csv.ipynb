# Formas de exportar CSV.

Una de las formas es usando la sentencia ***with*** y la funcion ***open()*** junto con el parametro mode de la funcion. Cuando se define el modo, se utiliza una de las siguientes opciones:

    'r' - read

    'w' - escribir

    'a' - añadir

    '+' - crear nuevo archivo

Añadiremos as file al final, para asignar el resultado a un nombre de variable. En este caso, la llamaremos data.

```python
import csv
with open("data_set/Unicorn_Companies.csv", mode='r') as file:
    # data = file.read() # lee el csv que llamamos file
    data = csv.DictReader(file) # transforma en diccionarios
    data = list(data) # crea una lista de diccionarios
#for fila in data:  recorre cada fila de data
#    print(fila)  imprime cada diccionario por fila
i=0
while i<10: # Recorremos los primeros 10 elementos
    print(data[i])
    i=i+1
```
